# Etappe 10: Prestatie op de steilste segmenten

**Activiteit:** Stubaier Höhenweg, etappe 10 (21 september 2026)

**Vraag:** Hoe presteer ik (qua hartslag en VAM) op het steilste stuk van 500 meter vergeleken met de gehele klim?

**Methode:**
1. Het hoogste punt van de etappe wordt automatisch gedetecteerd. Alles daarvoor beschouwen we als de 'hoofdklim'.
2. Via een glijdend venster scannen we de gehele beklimming om het segment van exact 500 meter te vinden waarin de meeste hoogtemeters worden overbrugd.
3. De VAM (stijgsnelheid) en hartslag van dit steilste segment worden direct vergeleken met de gemiddelden van de totale beklimming.

**Resultaat:** een afbeelding in de map `output/`.

## Imports

```python
from utils.berekeningen import bereken_steilste_segment
from utils.config import ANALYSE, PADEN
from utils.grafieken import teken_segment_profiel
from utils.inlezen import bereken_verstreken_tijd, lees_fit_bestand, schoon_fit_data
from utils.visualisatie import exporteer_afbeelding
```

## Instellingen

```python
ETAPPE_NR = 10
FIT_BESTAND = PADEN.fit_map / "20260921_stubaier_hohenweg_etappe_10.fit"
DOEL_AFSTAND_M = 500.0  # Lengte van het te scannen segment in meters
```

## Data inlezen

```python
ruwe_data = lees_fit_bestand(FIT_BESTAND)
df = schoon_fit_data(ruwe_data, cadans_factor=ANALYSE.wandel_cadans_factor)
df = bereken_verstreken_tijd(df)
```

## Berekening: Isolatie van de klim en het steilste segment

```python
df_klim, df_rest, df_segment, statistieken = bereken_steilste_segment(df, doel_afstand_m=DOEL_AFSTAND_M)

print("--- Resultaten Beklimming ---")
print(f"Totale VAM klim: {statistieken['klim']['vam']} m/u")
print(f"VAM steilste 500m: {statistieken['segment']['vam']} m/u")
```

## Grafiek

```python
fig = teken_segment_profiel(df_klim, df_rest, df_segment, statistieken, doel_afstand_m=DOEL_AFSTAND_M, etappe_nr=ETAPPE_NR)
```

## Exporteren

```python
exporteer_afbeelding(fig, bestandsnaam="20260921_segment_performance", formaat="liggend")
```
